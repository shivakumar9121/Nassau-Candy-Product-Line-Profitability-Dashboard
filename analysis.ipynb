## 4.7 Pareto Analysis (80/20 Rule)

### Business Question

Which products contribute to 80% of the company's total revenue and total profit?

```python
pareto_sales = (
    df_clean.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

pareto_sales.columns = ["Product", "Sales"]

pareto_sales["Cumulative Sales"] = pareto_sales["Sales"].cumsum()

pareto_sales["Cumulative %"] = (
    pareto_sales["Cumulative Sales"]
    / pareto_sales["Sales"].sum()
) * 100

pareto_sales
```

```python
threshold_index = pareto_sales[pareto_sales["Cumulative %"] >= 80].index[0]

pareto_80 = pareto_sales.iloc[:threshold_index + 1]

pareto_80
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    pareto_sales["Product"],
    pareto_sales["Sales"],
    label="Sales"
)

plt.plot(
    pareto_sales["Product"],
    pareto_sales["Cumulative %"],
    color="red",
    marker="o",
    label="Cumulative %"
)

plt.axhline(
    y=80,
    color="green",
    linestyle="--",
    label="80% Threshold"
)

plt.xticks(rotation=45, ha="right")

plt.ylabel("Sales / Cumulative %")

plt.title("Pareto Analysis of Product Sales")

plt.legend()

plt.tight_layout()

plt.show()
```

### Observation

- The top four products contribute approximately 76% of the company's total revenue.
- Including the fifth product increases the cumulative revenue contribution to over 90%.
- Revenue is highly concentrated in a small number of products.

### Business Insight

The company is highly dependent on its top-performing chocolate products. Any disruption in demand or supply for these products could significantly affect overall business performance.

### Recommendation

Maintain sufficient inventory and marketing support for the top-performing products. At the same time, develop strategies to improve the sales of lower-performing products to reduce dependence on a small product portfolio.
